````julia
# Write your Julia/JuMP code here.

using JuMP, HiGHS
Spiderman=Model(HiGHS.Optimizer)
@variable(Spiderman,3<=P1) #Process 1 Time/vials C made
@variable(Spiderman, 0<=P2) #Process 2 Time
@constraint(Spiderman,10<=3P1+P2) #Vials A made
@constraint(Spiderman, 5 <= P1+P2) #Vials B made
@objective(Spiderman, Min, 4*P1+P2) #cost

optimize!(Spiderman)

status = termination_status(Spiderman)
status == MOI.OPTIMAL || error("HiGHS stopped with status $(status).")

process_1_time = value(P1)
process_2_time = value(P2)
cost_value = objective_value(Spiderman)

println("Termination status: ", status)
println("Produce ", process_1_time, " hours of Process 1.")
println("Produce ", process_2_time, " hours of Process 2.")
println("Spiderman Produced ", 3*process_1_time+process_2_time, " of vial A.")
println("Spiderman Produced ", process_1_time+process_2_time, " of vial B.")
println("Spiderman Produced ", process_1_time, " of vial C.")
println("Total cost: \$", cost_value)
````

## Problem 3 — Planning a brewery's production

The 524 Brewing and Bottling Company produces light beer, pale ale, and malt liquor. It has 90 units of malt, 65 units of hops, and 80 units of yeast available. Resource requirements per gallon and profit per gallon are:

| Product | Malt (units/gallon) | Hops (units/gallon) | Yeast (units/gallon) | Profit (dollars/gallon) |
| :-- | --: | --: | --: | --: |
| Light beer | 2 | 3 | 2 | 2.50 |
| Pale ale | 2 | 2 | 1.5 | 2.00 |
| Malt liquor | 3 | 1.5 | 2 | 2.50 |

Any amount produced can be sold at the stated profit. Fractional gallons are allowed. There are no minimum production requirements initially.

### 3(a): Mathematical formulation

Formulate an LP to choose production quantities that maximize total profit. Define variables and units; state all resource constraints and variable domains.


**Formulation for 3(a):**

Type your response (LaTeX is welcome), or insert a photograph/scan of your handwritten response here. Include the required reasoning.

**decision variables:** \
    $LB$: Light Beer Produced (gallons)  $[0, +\infty)$\
    $PA$: Pale Ale Produced (gallons)  $[0, +\infty)$\
    $ML$: Malt Liquor Produced (gallons)  $[0,+\infty)$

**constraints:**\
    $2*LB+2*PA+3*ML \leq 90$ (Malt Constraint)\
    $3*LB+2*PA+1.5*ML \leq 65$ (Hops Constraint)\
    $2*LB+1.5*PA+2*ML \leq 80$ (Yeast Constraint)

**Objective**\
    Maximize $2.5*LB+2*PA+2.5*ML$ (Profit $)

### 3(b): Implement, solve, and interpret

Implement and solve your LP with HiGHS. Print the solver status, gallons of each product, total profit, and the amount of each resource used and left over.

Which products are produced, and which resources are fully used? Explain why having a positive profit per gallon does not ensure that a product will appear in an optimal plan. Relate your explanation to this instance rather than just stating a general rule.


````julia
# Write your Julia/JuMP code here.
using JuMP, HiGHS
Brew1=Model(HiGHS.Optimizer)
@variable(Brew1, 0<=LB) #Light Beer Produced (Gallons)
@variable(Brew1, 0<=PA) #Pale Ale Produced (Gallons)
@variable(Brew1, 0<=ML) #Malt Liquor Produced (Gallons)
@constraint(Brew1,2*LB+2*PA+3*ML<=90) #Malt Constraint
@constraint(Brew1,3*LB+2*PA+1.5*ML<=65) #Hops Constraint
@constraint(Brew1,2*LB+1.5*PA+2*ML<=80) #Yeast Constraint
@objective(Brew1, Max, 2.5*LB+2*PA+2.5*ML) #Profit
 
optimize!(Brew1)

status = termination_status(Brew1)
status == MOI.OPTIMAL || error("HiGHS stopped with status $(status).")

Light_Beer_Produced = value(LB)
Pale_Ale_Produced = value(PA)
Malt_Liquor_Produced = value(ML)
profit = objective_value(Brew1)

println("Termination status: ", status)
println("Produce ", Light_Beer_Produced, " gallons of Light Beer.")
println("Produce ", Pale_Ale_Produced, " gallons of Pale Ale.")
println("Produce ", Malt_Liquor_Produced, " gallons of Malt Liquor.")
println("This will use ", 2*Light_Beer_Produced+2*Pale_Ale_Produced+3*Malt_Liquor_Produced, " units of Malt")
println("This will have ", 90-(2*Light_Beer_Produced+2*Pale_Ale_Produced+3*Malt_Liquor_Produced), " units of Malt left")
println("This will use ", 3*Light_Beer_Produced+2*Pale_Ale_Produced+1.5*Malt_Liquor_Produced, " units of Hops")
println("This will have ", 65-(3*Light_Beer_Produced+2*Pale_Ale_Produced+1.5*Malt_Liquor_Produced), " units of Hops left")
println("This will use ", 2*Light_Beer_Produced+1.5*Pale_Ale_Produced+2*Malt_Liquor_Produced, " units of Yeast")
println("This will have ", 80-(2*Light_Beer_Produced+1.5*Pale_Ale_Produced+2*Malt_Liquor_Produced), " units of Yeast left")
println("Total profit: \$", profit)
````
