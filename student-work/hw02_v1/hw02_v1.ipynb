## Problem 2 — Mission Impossible: a general indexed model

Jacob's Meat Alternative Plant (JMAP) prepares plant-based substitutes for several pork products. A product $p$ in a set $P$ can be prepared and sold using a method $c$ in a set $C$.

For each product $p$, at most $b_p$ units are available each day. Method $c$ can process at most $u_c$ units per day across **all** products. Selling one unit of product $p$ prepared by method $c$ brings in $h_{pc}$ dollars **before** the processing cost $q_c$ dollars per unit is deducted.

One unit of a product uses one unit of processing capacity and produces one unit for sale. Each unit is assigned to at most one method. Fractional units are allowed in this continuous planning model. All product–method combinations in the data are allowed. Prepared products can all be sold at the stated prices. Product may be left unused with no salvage value or disposal cost; there are no minimum sales requirements. Other costs are fixed and do not affect this decision.

### 2(a): General mathematical formulation

Write an LP that maximizes JMAP's daily net profit. Define the decisions and their units. Use $P,C,b_p,u_c,h_{pc},q_c$ to write a **general indexed formulation** that works for any numbers of products and methods. Explain the meaning and direction of each family of constraints. Do not substitute the numerical data into this written model.


**Formulation and reasoning for 2(a):**

Type your answer and reasoning in Markdown, or insert a clear photograph/scan of handwritten work. Label the subparts.

Sets: P (products), C (Method of Sale)

Parameters: $b_p$ (units available per product), $u_c$ (units availble per process), $h_{pc}$ (sale gross profit), $q_c$ (sale costs per process)


Objective function sums net profit of every product across each process and maximizes it.
$$
\begin{aligned}
\max\quad \sum_{p \epsilon P}\sum_{c \epsilon C} (h_{pc}-q_c)x_{pc}\\
\end{aligned}
$$  

The first set of constraints sums the number of products per process and makes sure it is less than or equal to the limit of units that can be processed by that process.

$$
\begin{aligned}
\sum_{p \epsilon P}x_{pc}\leq u_c \quad \forall p \epsilon P\\
\end{aligned}
$$  

The second set of constraints sums the number of products processed and makes sure it is less than or equal to the limit of units are avalible to sell that day.

$$
\begin{aligned}
\sum_{c \epsilon C}x_{pc}\leq b_p \quad \forall c \epsilon C\\
\end{aligned}
$$  

The final constraint makes sure a non-negitive amount of products is processed

$$
\begin{aligned}
\ x_{pc} \geq 0 \quad \forall c \epsilon C, p\epsilon P\\
\end{aligned}
$$
