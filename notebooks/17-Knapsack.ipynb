# Knapsack problem

My knapsack holds at most 15kg. I have the following items:

|Item number | 1  | 2  | 3  | 4  | 5  |
|------------|----|----|----|----|----|
|weight (kg) | 12 | 2  | 4  | 1  | 1  |
|value ($)   | 4  | 2  | 10 | 2  | 1  |

How can I maximize the value of the items in my knapsack?

Open the course repository root in VS Code, select the Julia 1.12 kernel with the course environment, and choose **Run All**. All required packages are included; no external data files are needed.


### First formulation

Let $z_i=1$ if item $i$ is packed, and $z_i=0$ otherwise. Each item can be selected at most once.


```julia
using JuMP, HiGHS

m = Model(HiGHS.Optimizer)
set_silent(m)
@variable(m, z[1:5], Bin )
@constraint(m, 12z[1] + 2z[2] + 4z[3] + z[4] + z[5] <= 15)
@objective(m, Max, 4z[1] + 2z[2] + 10z[3] + 2z[4] + 1z[5])

optimize!(m)
if !is_solved_and_feasible(m; allow_local = false)
    error("No optimal primal solution: $(termination_status(m)), $(primal_status(m))")
end

println(termination_status(m))
println("z = ", value.(z) )
println("objective = ", objective_value(m) )
```

### General formulation

Suppose we have weights $w_1,\dots,w_n$ and a weight limit $W$. Suppose the items have values of $v_1,\dots,v_n$. How should we choose the items in order to maximize value while satisfying weight limit.

The indexed formulation is

$$
\begin{aligned}
\max\quad &\sum_{i=1}^n v_i z_i\\
\text{s.t.}\quad &\sum_{i=1}^n w_i z_i\le W,\\
&z_i\in\{0,1\} &&i=1,\ldots,n.
\end{aligned}
$$


```julia
# parameters for our problem
w = [12, 2, 4, 1, 1]  # weights
v = [4, 2, 10, 2, 1]  # values
W = 15                # weight limit
n = length(w);        # number of items
```

```julia
using JuMP, HiGHS
m = Model(HiGHS.Optimizer)
set_silent(m)

@variable(m, z[1:n], Bin )
@constraint(m, sum( w[i]*z[i] for i=1:n) <= W )
@objective(m, Max, sum( v[i]*z[i] for i=1:n) )

optimize!(m)
if !is_solved_and_feasible(m; allow_local = false)
    error("No optimal primal solution: $(termination_status(m)), $(primal_status(m))")
end

println(termination_status(m))
println("z = ", value.(z) )
println("objective = ", objective_value(m) )
```
