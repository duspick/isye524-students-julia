# Top Brass: an introductory linear program

Top Brass Trophy Company makes championship trophies for youth football and soccer leagues. Each football trophy earns $\$12$ in profit and uses 4 board feet of wood, one engraved plaque, and one brass football. Each soccer trophy earns $\$9$ in profit and uses 2 board feet of wood, one engraved plaque, and one brass soccer ball.

The company has 4,800 board feet of wood, 1,750 plaques, 1,000 brass footballs, and 1,500 brass soccer balls. Assuming every completed trophy can be sold, how many of each type should the company produce to maximize profit?

## The mathematical model

Let $f$ be the number of football trophies and $s$ the number of soccer trophies. The introductory linear-programming model is

$$
\begin{aligned}
\max_{f,s} \quad & 12f + 9s && \text{(profit)} \\
\text{s.t.} \quad & 4f + 2s \leq 4800 && \text{(wood)} \\
& f + s \leq 1750 && \text{(plaques)} \\
& f \leq 1000 && \text{(brass footballs)} \\
& s \leq 1500 && \text{(brass soccer balls)} \\
& f, s \geq 0.
\end{aligned}
$$

This LP treats production quantities as continuous. Its optimal quantities happen to be integers; later models will impose integrality explicitly when it is required.

```julia
using JuMP, HiGHS
import MathOptInterface as MOI

model = Model(HiGHS.Optimizer)
set_silent(model)

@variable(model, football_trophies >= 0)
@variable(model, soccer_trophies >= 0)

@objective(
    model,
    Max,
    12 * football_trophies + 9 * soccer_trophies,
)

@constraint(
    model,
    wood_limit,
    4 * football_trophies + 2 * soccer_trophies <= 4800,
)
@constraint(
    model,
    plaque_limit,
    football_trophies + soccer_trophies <= 1750,
)
@constraint(model, football_limit, football_trophies <= 1000)
@constraint(model, soccer_limit, soccer_trophies <= 1500)

model
```

## Solve with HiGHS

HiGHS is the open-source linear-programming solver included in the course environment. Always check the termination status before reading objective or variable values.

```julia
optimize!(model)

status = termination_status(model)
status == MOI.OPTIMAL || error("HiGHS stopped with status $(status).")

football_value = value(football_trophies)
soccer_value = value(soccer_trophies)
profit_value = objective_value(model)

println("Termination status: ", status)
println("Build ", football_value, " football trophies.")
println("Build ", soccer_value, " soccer trophies.")
println("Total profit: \$", profit_value)
```

## Interpret the solution

The optimal plan makes 650 football trophies and 1,100 soccer trophies for a total profit of $\$17{,}700$. It uses all 4,800 board feet of wood and all 1,750 plaques. The available brass footballs and soccer balls are not fully used.
